```python
from src.Optimization import OptimizationConfig, OptunaStreamOptimizer

selectedFeatures = [
    "Flow_Duration",
    "Total_Fwd_Packets",
    "Total_Backward_Packets",
    "Total_Length_of_Fwd_Packets",
    "Total_Length_of_Bwd_Packets",
    "Fwd_Packet_Length_Max",
    "Fwd_Packet_Length_Min",
    "Fwd_Packet_Length_Mean",
    "Bwd_Packet_Length_Max",
    "Bwd_Packet_Length_Mean",
    "Flow_Bytes_s",
    "Flow_IAT_Mean",
    "Flow_IAT_Std",
    "Flow_IAT_Max",
    "Flow_IAT_Min",
    "Fwd_IAT_Total",
    "Fwd_IAT_Mean",
    "Fwd_IAT_Std",
    "Fwd_IAT_Max",
    "Bwd_Packets_s",
    "Min_Packet_Length",
    "Max_Packet_Length",
    "Packet_Length_Mean",
    "ACK_Flag_Count",
    "URG_Flag_Count",
    "Down_Up_Ratio",
    "Average_Packet_Size",
    "Avg_Fwd_Segment_Size",
    "Subflow_Fwd_Packets",
    "Subflow_Fwd_Bytes",
    "Subflow_Bwd_Bytes",
    "Init_Win_bytes_forward",
    "act_data_pkt_fwd",
    "min_seg_size_forward",
    "SourceIPEntropy",
    "DestinationIPEntropy",
    "SourcePortEntropy",
    "DestinationPortEntropy",
    "ProtocolEntropy",
]

task = "anomaly"  # Use "classification" para otimizar classificadores.
experimentName = "CICDDoS2019_Optimization"

datasets = {
    "Consistency_1s": {
        "path": r"data/attack scenario window/Consistency_1s.csv",
        "selectedFeatures": selectedFeatures,
        "removedFeatures": ["Timestamp", "AttackRatio"],
    },
}
```

```python
if task == "anomaly":
    parameters = {
        "model": {"code": "AIF"},
        "preprocessing": {
            "imputer": "incrementalMean",
            "normalizer": "incrementalZScore",
        },
        "training": {"strategy": "all"},
    }
else:
    parameters = {
        "model": {"code": "HAT"},
        "preprocessing": {
            "imputer": "incrementalMean",
            "normalizer": "incrementalZScore",
        },
    }

config = OptimizationConfig(
    experimentName=experimentName,
    datasets=datasets,
    parameters=parameters,
    task=task,
    nTrials=100,
    optunaSeed=42,
)
```

```python
optimizationResult = OptunaStreamOptimizer(
    config=config,
).run()

display(optimizationResult["summary"])
```
