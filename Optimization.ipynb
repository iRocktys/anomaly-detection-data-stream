# Optimization
Otimização binária com Optuna para detectores de anomalia ou classificadores. F1 é a única função objetivo.


```python
from pathlib import Path

from ProjectDefaults import getDefaultParameters
from src.Optimization import OptimizationConfig, OptunaStreamOptimizer
```

```python
task = "anomaly"  # Use "classification" para otimizar classificadores.
experimentName = "Optimization_1s"

datasets = {
    "Consistency_1s": {"path": Path("output/Scenarios/CICDDoS2019/Datasets/Consistency_1s.csv")},
}

parameters = getDefaultParameters(task)
if task == "anomaly":
    parameters["model"]["code"] = "AIF"
else:
    parameters["model"]["code"] = "HAT"
```

```python
config = OptimizationConfig(
    experimentName=experimentName,
    task=task,
    datasets=datasets,
    parameters=parameters,
    nTrials=100,
    optimizeModelParameters=True,
)

optimizationResult = OptunaStreamOptimizer(config).run()
display(optimizationResult["summary"])
```
