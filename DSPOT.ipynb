# DSPOT Study

```python
import pandas as pd
from src.DSPOT import DspotStudy
from src.Thresholds.Incremental.DspotThreshold import DspotConfig
```

```python
scorePath = r"output\Optimization\CICDDoS2019_Optimization\AIF\Consistency_1s\best\instances.csv"
scoreColumn = "score"
experimentName = "DSPOT_Study"

scoreFrame = pd.read_csv(scorePath)

print("Quantidade de scores carregados:", len(scoreFrame))
```

```python
# Configuração do estudo de otimização (Consistência 1s - AIF )
config = DspotConfig(
    driftDepth=20,
    calibrationSize=480,
    initialQuantile=0.9681225931442474,
    risk=0.0007412528694192007,
    refitEvery=13,
    optimizationStarts=20,
    tolerance=8.275861144192933e-09,
)
```

```python
studyResult = DspotStudy().run(
    experimentName=experimentName,
    scores=scoreFrame.iloc[500:].reset_index(drop=True), # Remove as primeiras 500 linhas de aquecimento do AIF, para gerar graficos identicos aos do estudo de otimização
    scoreColumn=scoreColumn,
    config=config,
)

display(studyResult["resultFrame"])
print("Gráfico salvo em:", studyResult["plotPath"])
```
