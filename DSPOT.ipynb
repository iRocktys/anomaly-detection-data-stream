# DSPOT
Estudo do DSPOT utilizando a mesma implementação de produção usada em `AnomalyDetection`.


```python
from pathlib import Path
import pandas as pd
from src.DSPOT import DspotStudy
from src.Thresholds.Incremental.DspotThreshold import DspotConfig
```

```python
scorePath = Path("data\\BALANCED.CSV")
scoreColumn = "score"
experimentName = "DSPOT_Study"

scoreFrame = pd.read_csv(scorePath)
scores = pd.to_numeric(scoreFrame[scoreColumn], errors="coerce").dropna().to_numpy()
```

```python
config = DspotConfig(
    driftDepth=50,
    calibrationSize=450,
    initialQuantile=0.95,
    risk=0.001,
    refitEvery=10,
    optimizationStarts=10,
    tolerance=1e-8,
)

studyResult = DspotStudy().run(experimentName, scores, config)
display(studyResult["resultFrame"].tail())
```
