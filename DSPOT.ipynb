# DSPOT Study

This notebook studies the same `DspotThreshold` implementation used by anomaly-detection experiments.


```python
import pandas as pd

from src.DSPOT import DspotStudy
from src.Thresholds.Incremental.DspotThreshold import DspotConfig
```

```python
scorePath = r"output\CICDDoS2019_Window_1s\AIF\Consistency_1s\001-ALL-DSPOT-MA25-MEAN\instances.csv"
scoreColumn = "scoreMa25"
experimentName = "DSPOT_Study"

scoreFrame = pd.read_csv(scorePath)
scores = pd.to_numeric(scoreFrame[scoreColumn], errors="coerce").dropna().to_numpy()

print("Quantidade de scores carregados:", len(scores))
```

```python
config = DspotConfig(
    driftDepth=20,
    calibrationSize=480,
    initialQuantile=0.9681225931442474,
    risk=0.0007412528694192,
    refitEvery=13,
    optimizationStarts=20,
    tolerance=8.275861144192933e-09,
)
```

```python
studyResult = DspotStudy().run(
    experimentName=experimentName,
    scores=scores,
    config=config,
)

display(studyResult["resultFrame"].tail())
print("Gráfico salvo em:", studyResult["plotPath"])
```
